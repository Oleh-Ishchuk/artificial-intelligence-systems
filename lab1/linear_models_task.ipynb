**Вправа 14.** Знайди $x$, який мінімізує $(x-3)^2 + 5$ — не аналітично, а градієнтним спуском: `x` як параметр з `requires_grad=True`, без жодної моделі.

````python
x = torch.tensor(0.0, requires_grad=True)
lr = 0.1

# TODO: цикл градієнтного спуску (20 кроків):
#   loss = (x - 3) ** 2 + 5
#   loss.backward()
#   під torch.no_grad(): x -= lr * x.grad
#   x.grad.zero_()
for step in range(20):
    loss = (x - 3) ** 2 + 5
    loss.backward()
    with torch.no_grad():
        x -= lr * x.grad
    x.grad.zero_()

print(f"x = {x.item():.4f}  (мінімум у x=3)")
````

## Завдання 1. Лінійна регресія

Тренуємо базовий пайплайн на простій лінійній регресії з одним параметром.

$$y = w \cdot x + b + \varepsilon, \quad \varepsilon \sim \mathcal{N}(0, 1)$$

````python
TASK1_TRUE_W = 2.0
TASK1_TRUE_B = 1.0
TASK1_N = 300

x1 = torch.empty(TASK1_N, 1).uniform_(-5, 5)
noise = 1.0 * torch.randn(TASK1_N, 1)
y1 = TASK1_TRUE_W * x1 + TASK1_TRUE_B + noise
````

**Вправа 17.** Перш ніж будувати модель — подивись на самі дані.

````python
# TODO: Точкова діаграма (scatter plot) з вісями x1 та y1, підпис осей (xlabel/ylabel), заголовок
# "Сирі дані (до тренування)"
plt.scatter(x1.numpy(), y1.numpy(), s=10, alpha=0.6)
plt.xlabel("x1")
plt.ylabel("y1")
plt.title("Сирі дані (до тренування)")
plt.grid(alpha=0.3)
plt.show()
````

**Вправа 18.** Розбий дані на train/val/test (70/15/15). Перемішай дані перед зрізом, інакше спліт вийде не випадковим.

````python
# TODO: перемішати x1/y1 через torch.randperm(TASK1_N), розбити на
# train/val/test у пропорції 70/15/15
# x_train1, y_train1 = ...
# x_val1, y_val1 = ...
# x_test1, y_test1 = ...

perm = torch.randperm(TASK1_N)
x1_shuffled = x1[perm]
y1_shuffled = y1[perm]

n_train = int(0.7 * TASK1_N)
n_val = int(0.15 * TASK1_N)

x_train1, y_train1 = x1_shuffled[:n_train], y1_shuffled[:n_train]
x_val1, y_val1 = x1_shuffled[n_train:n_train + n_val], y1_shuffled[n_train:n_train + n_val]
x_test1, y_test1 = x1_shuffled[n_train + n_val:], y1_shuffled[n_train + n_val:]
print(f"train/val/test: {len(x_train1)}/{len(x_val1)}/{len(x_test1)}")
````

**Вправа 19.** Нижче — готовий цикл навчання, кожен крок підписаний окремо. Просто запусти обидві комірки й подивись, як модель наближається до `true` за 100 епох (знімки кожні 30 епох).

````python
task1_model = torch.nn.Linear(in_features=1, out_features=1)
optimizer = torch.optim.SGD(task1_model.parameters(), lr=0.05)
loss_fn = torch.nn.MSELoss()

x_line = torch.linspace(-5, 5, 100).unsqueeze(1)  # сітка для візуалізації лінії
snapshots = []  # (epoch, w, b, y_line_pred) кожні 30 епох

for epoch in range(100):
    # обнулити градієнти з попереднього кроку
    optimizer.zero_grad()
    # forward -- прогноз моделі на поточних w, b
    y_pred = task1_model(x_train1)
    # loss -- наскільки прогноз відрізняється від y_train1
    loss = loss_fn(y_pred, y_train1)
    # backward -- градієнти loss за w і b
    loss.backward()
    # крок оптимізатора -- оновити w і b
    optimizer.step()

    if epoch % 30 == 0 or epoch == 99:
        with torch.no_grad():
            y_line_pred = task1_model(x_line)
        snapshots.append((epoch, task1_model.weight.item(), task1_model.bias.item(), y_line_pred))
        print(f"epoch {epoch:4d}  loss={loss.item():.6f}  "
              f"w={task1_model.weight.item():.4f} (true {TASK1_TRUE_W})  "
              f"b={task1_model.bias.item():.4f} (true {TASK1_TRUE_B})")
````

````python
with torch.no_grad():
    y_line_true = TASK1_TRUE_W * x_line + TASK1_TRUE_B

fig, axes = plt.subplots(1, len(snapshots), figsize=(4 * len(snapshots), 3.5), sharey=True)
for ax, (epoch, w, b, y_line_pred) in zip(axes, snapshots):
    ax.scatter(x1.numpy(), y1.numpy(), s=8, alpha=0.3, label="дані")
    ax.plot(x_line.numpy(), y_line_true.numpy(), "g--", label="true")
    ax.plot(x_line.numpy(), y_line_pred.numpy(), "r-", label="модель")
    ax.set_title(f"epoch {epoch}\nw={w:.2f}, b={b:.2f}")
    ax.grid(alpha=0.3)
axes[0].legend()
plt.tight_layout()
plt.show()
````

### Сценарій: рекламний бюджет → кількість продажів

Я обрав залежність між тижневим бюджетом на рекламу (тис. грн) і
кількістю продажів товару за тиждень.

Закладена залежність: y = 3·x + 15 + ε, де ε ~ N(0, 4)

- **w = 3** — кожна додаткова тисяча гривень у рекламу приносить
  у середньому +3 продажі
- **b = 15** — базовий рівень продажів без будь-якої реклами
  (органічний попит)
- Діапазон бюджету: 0–20 тис. грн/тиждень — реалістично для
  малого бізнесу
- Шум N(0, 4) відображає коливання попиту, які реклама не пояснює
  (сезонність, конкуренти тощо)

**Чому нормалізація ознаки.** На відміну від Завдання 1 (де x був
у межах [-5, 5], а y — приблизно того ж порядку), тут x (0–20) і
y (15–75) мають різний масштаб. Якщо тренувати без змін, градієнт
за w виявляється набагато більшим за градієнт за b (бо в формулу
градієнта w входить множник x), тому w навчається швидко, а b —
у рази повільніше: за 300 епох b встигає доїхати лише до ~3
замість 15 (перевірено окремим прогоном).

Щоб вирішити це, нормалізую x (z-score, як у Вправі 6: віднімаю
середнє і ділю на std, порахований лише на train-вибірці). Це
вирівнює масштаб градієнтів для w і b, і модель навчається так
само швидко й стабільно, як у Завданні 1. Після навчання
переводжу параметри назад у "справжній" масштаб (грн), щоб
порівняти з true w і b.

### Генерація даних ###

````python
TASK2_TRUE_W = 3.0
TASK2_TRUE_B = 15.0
TASK2_N = 300
TASK2_NOISE_STD = 4.0
TASK2_X_MIN, TASK2_X_MAX = 0.0, 20.0

def generate_ads_sales_data(n, w, b, x_min, x_max, noise_std):
    """x -- рекламний бюджет (тис. грн), y -- к-сть продажів за тиждень."""
    x = torch.empty(n, 1).uniform_(x_min, x_max)
    noise = noise_std * torch.randn(n, 1)
    y = w * x + b + noise
    return x, y

x2, y2 = generate_ads_sales_data(TASK2_N, TASK2_TRUE_W, TASK2_TRUE_B,
                                   TASK2_X_MIN, TASK2_X_MAX, TASK2_NOISE_STD)
````

### Збереження датасету в CSV

Зберігаю згенеровані дані (`x2`, `y2`) у CSV файл на Google Drive
(`dataset.csv`).

Колонки: `ad_budget_k_uah` (бюджет у тис. грн) і `sales`
(кількість продажів).

````python
def save_dataset_to_csv(x, y, path):
    df_out = pd.DataFrame({
        "ad_budget_k_uah": x.squeeze().numpy(),
        "sales": y.squeeze().numpy()
    })
    df_out.to_csv(path, index=False)
    return df_out

dataset_path = os.path.join(ARTIFACT_DIR, "dataset.csv")
save_dataset_to_csv(x2, y2, dataset_path)
print("Збережено в:", dataset_path)
````

### Розбиття на train/val/test

Перемішую дані (`torch.randperm`) і ріжу на три частини у
пропорції 70/15/15, як і в Завданні 1 — окрема train-вибірка для
навчання, val для контролю під час тренування, test для фінальної
незалежної перевірки.

````python
def split_dataset(x, y, train_frac=0.7, val_frac=0.15):
    n = x.shape[0]
    perm = torch.randperm(n)
    x, y = x[perm], y[perm]
    n_train = int(train_frac * n)
    n_val = int(val_frac * n)
    return (x[:n_train], y[:n_train]), \
           (x[n_train:n_train+n_val], y[n_train:n_train+n_val]), \
           (x[n_train+n_val:], y[n_train+n_val:])

(x_train2, y_train2), (x_val2, y_val2), (x_test2, y_test2) = split_dataset(x2, y2)
print(f"train/val/test: {len(x_train2)}/{len(x_val2)}/{len(x_test2)}")
````

### Нормалізація ознаки

На відміну від Завдання 1 (де x був у межах [-5, 5], близько до
масштабу y), тут x (0–20) і y (15–75) мають різний масштаб. Без
нормалізації градієнт за w виявляється набагато більшим за
градієнт за b, тому w навчається швидко, а b — у рази повільніше.

Щоб це виправити, нормалізую x (z-score, як у Вправі 6): віднімаю
середнє і ділю на std, пораховані лише на train-вибірці (щоб не
"підглядати" в val/test). Це вирівнює масштаб градієнтів, і модель
навчається так само швидко, як у Завданні 1.

````python
def normalize_feature(x_train, x_val, x_test):
    mean, std = x_train.mean(), x_train.std()
    return (x_train - mean) / std, (x_val - mean) / std, (x_test - mean) / std, mean, std

x_train2_n, x_val2_n, x_test2_n, x2_mean, x2_std = normalize_feature(x_train2, x_val2, x_test2)
````

### Тренування моделі

Той самий цикл навчання, що й у Завданні 1 (SGD + MSELoss),
оформлений у перевикористовувану функцію. Тренуємось на
нормалізованому x, а після навчання переводимо параметри назад у
"справжній" масштаб (тис. грн), щоб порівняти з true w і b.

````python
def train_linear_model(x_train, y_train, x_val, y_val, epochs=100, lr=0.05):
    model = torch.nn.Linear(1, 1)
    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    loss_fn = torch.nn.MSELoss()

    for epoch in range(epochs):
        optimizer.zero_grad()
        y_pred = model(x_train)
        loss = loss_fn(y_pred, y_train)
        loss.backward()
        optimizer.step()

        if epoch % 20 == 0 or epoch == epochs - 1:
            with torch.no_grad():
                val_loss = loss_fn(model(x_val), y_val)
            print(f"epoch {epoch:4d}  train_loss={loss.item():.4f}  val_loss={val_loss.item():.4f}")
    return model

task2_model = train_linear_model(x_train2_n, y_train2, x_val2_n, y_val2, epochs=100, lr=0.05)

w_norm, b_norm = task2_model.weight.item(), task2_model.bias.item()
w_real = w_norm / x2_std.item()
b_real = b_norm - w_norm * x2_mean.item() / x2_std.item()
print(f"\nНавчено: w={w_real:.4f} (true {TASK2_TRUE_W}), b={b_real:.4f} (true {TASK2_TRUE_B})")
````

### Перевірка на test

Рахуємо MSE на відкладеній test-вибірці, яку модель ще не бачила
під час навчання — це фінальна незалежна оцінка якості моделі.

````python
def evaluate_model(model, x_test, y_test):
    loss_fn = torch.nn.MSELoss()
    with torch.no_grad():
        test_loss = loss_fn(model(x_test), y_test)
    return test_loss.item()

test_loss2 = evaluate_model(task2_model, x_test2_n, y_test2)
print("Test loss (MSE):", test_loss2)
````
